```python
import os
import shutil

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import random as rd
import time

if os.path.exists('generated_graphs'):
    shutil.rmtree('generated_graphs')  # Supprime tout le dossier et son contenu
os.makedirs('generated_graphs', exist_ok=True)  # Recrée un répertoire vide


def generate_random_tree(n):
    return nx.random_labeled_tree(n)


# Fonction pour obtenir la matrice d'adjacence
def adjacency(G): # La fonction de nx donne la même matrice à isomorphisme près

    n=len(G.nodes())

    matrice=np.zeros((n,n))

    for e in list(G.edges()):
        matrice[e]=1
        matrice[e[1],e[0]]=1

    return matrice

def dessiner(G):

    # pos = nx.nx_agraph.graphviz_layout(G, prog="dot")  # Nécessite pygraphviz ou pydot

    plt.figure(figsize=(8, 6))
    nx.draw(
        G, with_labels=True, node_color="lightblue", edge_color="gray",
        node_size=2000, font_size=12, font_weight="bold"
    )
    plt.title("Arbre")
    plt.show()


# Générateur (Génère une matrice d'adjacence)
class GraphGenerator(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(GraphGenerator, self).__init__()
        self.fc = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, output_dim),
            nn.Sigmoid()
        )

    def forward(self, z):
        return self.fc(z)

# Discriminateur (Juge si une matrice d'adjacence est réelle ou générée)
class GraphDiscriminator(nn.Module):
    def __init__(self, input_dim):
        super(GraphDiscriminator, self).__init__()
        self.fc = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 1),
            nn.Sigmoid()
        )

    def forward(self, graph):
        return self.fc(graph)



# Enregistrer les graphes dans le dataset
def save_graphs(graph_dataset, n_graphs, n_nodes):
    for i in range(n_graphs):
        adj_matrix = graph_dataset[i].reshape(n_nodes, n_nodes)
        G = nx.from_numpy_array(adj_matrix)

        # Créer un fichier d'image pour chaque graphe
        plt.figure(figsize=(8, 6))
        try:
            cycles=nx.cycle_basis(G)

            if ( (cycles==[]) and (G.number_of_edges()==n_nodes-1) ): # Vérifie si le graphe est un arbre
                continue
                dessiner(G)

            elif ((set(cycles[0])==set(G.nodes)) and (len(cycles)==1)):# Vérifie si un gaphe est un graphe cycle.
                pos = nx.kamada_kawai_layout(G) # For draw like a regular polygon
                nx.draw(G,pos, with_labels=True, node_color="lightblue", font_weight="bold", node_size=700, edge_color="gray")

            else:
                nx.draw_planar(G, with_labels=True, node_color="lightblue", font_weight="bold", node_size=700, edge_color="gray")


        except nx.NetworkXException:
            # Si le graphe n'est pas planaire, utiliser un autre layout
            nx.draw(G,with_labels=True, node_color="lightblue", font_weight="bold", node_size=700, edge_color="gray")

        plt.title(f"Graphe {i + 1}")
        plt.show()

        # Enregistrer l'image dans le dossier 'generated_graphs'
        plt.savefig(f'generated_graphs/graph_{i + 1}.png')
        plt.close()  # Fermer la figure pour libérer de la mémoire

def train(n_nodes=10,n_graphs=100, fct=generate_random_tree, num_epochs=10_000, batch_size=32, num_graphe_GAN=10):
    """
    Entraîne un GAN pour générer des graphes.

    Args:
        n_nodes (int): Nombre de sommets, 10 par défaut.
        n_graphs (int): Nombre de graphes générés pour l'entraînement, 100 par défaut.
        fct (function): Fonction utilisée pour générer les graphes. Les options incluent:
            - generate_random_binary_tree() (fonctionne bien)
            - generate_random_bipartite_graph (fonctionne bien)
            - generate_random_planar_graph (ne fonctionne pas encore bien)
            - generate_random_cyclic_graph (fonctionne bien)
            - generate_random_cycle_graph (ne fonctionne pas encore bien)
            - generate_random_tree (fonctionne bien)
        num_epochs (int): Nombre d'époques pour l'entraînement, 10_000 par défaut.
        batch_size (int): Taille des lots pour l'entraînement, 32 par défaut.
        num_graphe_GAN (int): Nombre de graphes générés par le GAN à la fin de l'entraînement, 10 par défaut.
    """


    adj_matrix_dim = n_nodes * n_nodes  # Représentation sous forme de matrice d'adjacence

    # Créer les modèles
    generator = GraphGenerator(input_dim=adj_matrix_dim, output_dim=adj_matrix_dim)
    discriminator = GraphDiscriminator(input_dim=adj_matrix_dim)

    g_optimizer = optim.Adam(generator.parameters(), lr=0.001)
    d_optimizer = optim.Adam(discriminator.parameters(), lr=0.001)

    loss_fn = nn.BCELoss()


    graph_dataset = np.array([adjacency(fct(n_nodes)) for _ in range(n_graphs)])
    graph_dataset = graph_dataset.reshape(n_graphs, -1)
    # Enregistrer les graphes dans le dataset
    save_graphs(graph_dataset, n_graphs, n_nodes)

    d_losses = []
    g_losses = []

    for epoch in range(num_epochs):
        # Étape 1: Entraîner le Discriminateur avec des graphes réels et générés
        idx = np.random.randint(0, n_graphs, batch_size)
        real_graphs = torch.tensor(graph_dataset[idx], dtype=torch.float32)  # Sélection de graphes réels
        z = torch.randn(batch_size, adj_matrix_dim)  # Bruit aléatoire pour le générateur
        fake_graphs = generator(z)

        real_labels = torch.ones(batch_size, 1)
        fake_labels = torch.zeros(batch_size, 1)

        d_optimizer.zero_grad()
        real_loss = loss_fn(discriminator(real_graphs), real_labels)
        fake_loss = loss_fn(discriminator(fake_graphs.detach()), fake_labels)
        d_loss = real_loss + fake_loss
        d_loss.backward()
        d_optimizer.step()

        # Étape 2: Entraîner le Générateur pour tromper le Discriminateur
        g_optimizer.zero_grad()
        generated_loss = loss_fn(discriminator(fake_graphs), real_labels)  # Le générateur essaie de tromper le discriminateur
        generated_loss.backward()
        g_optimizer.step()


        d_losses.append(d_loss.item())
        g_losses.append(generated_loss.item())

        # Afficher la loss toutes les 500 époques
        if epoch % 500 == 0:
            print(f"Epoch [{epoch}/{num_epochs}], Loss D: {d_loss.item()}, Loss G: {generated_loss.item()}")
    plt.figure(figsize=(10, 5))
    plt.plot(d_losses, label='Loss du discriminator')
    plt.plot(g_losses, label='Loss du generator')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.title('Loss du GAN')
    plt.show()
    for _ in range(num_graphe_GAN):
        # Afficher les graphes générés à la fin
        z = torch.randn(1, adj_matrix_dim)  # Bruit aléatoire
        generated_adj_matrix = generator(z).view(n_nodes, n_nodes).detach().numpy()

        threshold = 0.5
        binary_adj_matrix = (generated_adj_matrix > threshold).astype(int)

        # Création du graphe à partir de la matrice d'adjacence
        G = nx.from_numpy_array(binary_adj_matrix)

        plt.figure(figsize=(8, 6))
        try:
            cycles=nx.cycle_basis(G)



            if ( (cycles==[]) ): # Vérifie si le graphe est acyclique.    un arbre
                dessiner(G)

            elif ((set(cycles[0])==set(G.nodes)) and (len(cycles)==1)):# Vérifie si un gaphe est un graphe-cycle.
                pos = nx.kamada_kawai_layout(G) # For draw like a regular polygon
                nx.draw(G,pos, with_labels=True, node_color="lightblue", font_weight="bold", node_size=700, edge_color="gray")

            else:
                nx.draw_planar(G, with_labels=True, node_color="lightblue", font_weight="bold", node_size=700, edge_color="gray")


        except nx.NetworkXException:
            # Si le graphe n'est pas planaire, utiliser un autre layout
            nx.draw(G,with_labels=True, node_color="lightblue", font_weight="bold", node_size=700, edge_color="gray")
        plt.title(f"Graphe généré par le GAN après {num_epochs} époques")
        plt.show()  # Afficher le dernier graphe généré

train(n_graphs=100,n_nodes=15,fct=generate_random_tree, num_epochs=10_000, num_graphe_GAN=10)
```

```python
def generate_random_binary_tree(num_nodes): # Algo de Rémy. Pour aléatoire uniforme : Générer uniformément un mot de Dyck et le transformer en AB.
    """
    Génère un arbre binaire aléatoire avec une probabilité uniforme.

    Arguments :
        num_nodes (int): Nombre de nœuds dans l'arbre.

    Retourne :
        G (nx.DiGraph): Un graphe orienté représentant un arbre binaire.
    """
    if num_nodes < 1:
        raise ValueError("Un arbre binaire doit avoir au moins 1 nœud.")

    # Création récursive de l'arbre
    def build_tree(root, available_nodes):
        if not available_nodes:
            return []
        if len(available_nodes) == 1:
            return [(root, available_nodes[0])]

        # Choisir aléatoirement la taille du sous-arbre gauche
        left_size = rd.randint(0, len(available_nodes) - 1)
        left_nodes = available_nodes[:left_size]
        right_nodes = available_nodes[left_size:]

        # Assigner le prochain enfant gauche et droit (si disponibles)
        edges = []
        if left_nodes:
            left_child = left_nodes[0]
            edges.append((root, left_child))
            edges.extend(build_tree(left_child, left_nodes[1:]))

        if right_nodes:
            right_child = right_nodes[0]
            edges.append((root, right_child))
            edges.extend(build_tree(right_child, right_nodes[1:]))

        return edges

    # Initialiser la liste des nœuds
    nodes = list(range(num_nodes))
    root = nodes[0]  # Le premier nœud est la racine

    # Construire les arêtes de l'arbre binaire
    edges = build_tree(root, nodes[1:])

    # Créer un graphe NetworkX orienté
    G = nx.DiGraph()
    G.add_edges_from(edges)
    return G


train(n_graphs=100,n_nodes=10,fct=generate_random_binary_tree, num_epochs=10_000, num_graphe_GAN=10)
```
